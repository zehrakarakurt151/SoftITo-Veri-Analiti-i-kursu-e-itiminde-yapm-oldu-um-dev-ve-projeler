## Bölüm 4

* Veri kümesinde her bir Tür'ün (Genre) kaç kez geçtiğini gösteren bir pasta (pie) grafiği oluşturun.  
* En üretken 10 yazarı bulun ve kitap sayılarını çubuk grafik (bar plot) olarak görselleştirin.  
* Kullanıcı puanlarının (rating) dağılımını gösteren bir histogram grafiği oluşturun.


```python
import matplotlib.pyplot as plt
df["Genre"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Kitap Türlerinin Dağılımı")
plt.ylabel("")
plt.show()
```

```python
df["Author"].value_counts().head(10).plot(kind="bar")
plt.title("En Ürertken 10 Yazar")
plt.xlabel("Yazar")
plt.ylabel("Kitap Sayısı")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
df["User Rating"].plot(
    kind="hist",
    bins=10,
    edgecolor="black"
)

plt.title("Kullanıcı Puanlarının Dağılımı")
plt.xlabel("Kullanıcı Puanı")
plt.ylabel("Frekans")
plt.show()
```
